```python
from pyspark.sql.functions import udf
from pyspark.sql.types import ArrayType, FloatType
from openai import AzureOpenAI

# Pass credentials explicitly
API_KEY = os.environ["AZURE_OPENAI_API_KEY"]
ENDPOINT = os.environ["AZURE_OPENAI_ENDPOINT"]

def get_embedding_udf(text):
    client = AzureOpenAI(
        api_key=API_KEY,
        api_version="2024-02-01",
        azure_endpoint=ENDPOINT
    )
    
    response = client.embeddings.create(
        input=text,
        model="text-embedding-3-small"
    )
    
    return response.data[0].embedding

embedding_udf = udf(get_embedding_udf, ArrayType(FloatType()))
```

####applied the embedding function and created the vectors and created chunk_id which behaves like a primary key which is required by vector search index

```python
df_with_embeddings = df_chunked.withColumn(
    "embedding",
    embedding_udf("chunk_text")
).withColumn("chunk_id", monotonically_increasing_id())

display(df_with_embeddings)
```
